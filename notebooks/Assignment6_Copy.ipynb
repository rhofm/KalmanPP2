```python
# 1-D Gaussian blob
f = 8
g = 8
B = 2
center = 4
start_val_u = .06
sigma = 1.

def gaussian_blob_1d(f, g, center, sigma, amp, axis='col'):
    if axis == 'col':          # varies across columns, same down each row
        idx = np.arange(g)
        profile = amp * np.exp(-(idx-center)**2 / (2*sigma**2))
        return np.tile(profile, (f, 1))          # shape (f, g)
    else:                       # varies down rows, same across each column
        idx = np.arange(f)
        profile = amp * np.exp(-(idx-center)**2 / (2*sigma**2))
        return np.tile(profile, (g, 1)).T         # shape (f, g)

u0 = gaussian_blob_1d(f, g, center=center, sigma=sigma, amp=sart_val_u, axis='col')
a0 = B*u0

i0 = np.concatenate((u0.ravel(), a0.ravel()))
```

```python
# 2-D Gaussian blob
f = 8
g = 8
B = 2
center_x = 3
center_y = 5
start_val_u = .06
sigma = 1.

def gaussian_blob_2d(f, g, cx, cy, sigma, amp):
    y, x = np.mgrid[0:f, 0:g]
    return amp * np.exp(-((x-cx)**2 + (y-cy)**2) / (2*sigma**2))

u0 = gaussian_blob_2d(f, g, cx=center_x, cy=center_y, sigma=sigma, amp=sart_val_u)
a0 = B*u0

i0 = np.concatenate((u0.ravel(), a0.ravel()))
```
